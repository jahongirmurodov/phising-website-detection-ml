# Техническое задание

## Проект: Обнаружение фишинговых веб-страниц

---

## 1. Цель проекта

Разработать модель машинного обучения для **бинарной классификации веб-страниц**:

* `1` — фишинговый сайт
* `0` — безопасный сайт

Модель должна предсказывать класс страницы на основе структурных и технических характеристик URL и веб-страницы.

---

## 2. Бизнес-контекст

Фишинговые сайты используются для кражи личных данных пользователей (логины, пароли, банковские карты).
Разрабатываемая модель может быть использована:

* в браузерных расширениях
* в системах антифишинговой защиты
* в корпоративных системах кибербезопасности

---

## 3. Описание данных

Датасет содержит:

* признаки, описывающие URL и поведение страницы
* целевую переменную (phishing / legitimate)

## URL-структура

Характеризуют подозрительность URL.

* `length_url` — длина URL
* `length_hostname` — длина домена
* `ip` — используется ли IP вместо домена
* `nb_dots` — количество точек
* `nb_hyphens` — количество дефисов
* `nb_at` — наличие @
* `nb_slash` — количество слешей
* `ratio_digits_url` — доля цифр в URL
* `shortening_service` — используется ли сокращатель ссылок

Длинные и перегруженные URL чаще встречаются у фишинга.

---

## Доменная репутация

* `whois_registered_domain` — зарегистрирован ли домен
* `domain_registration_length` — срок регистрации
* `domain_age` — возраст домена
* `dns_record` — наличие DNS
* `google_index` — индексируется ли в Google
* `page_rank` — рейтинг страницы
* `web_traffic` — посещаемость

Фишинговые сайты часто новые и без репутации.

---

## HTML и поведение страницы

* `iframe` — есть ли iframe
* `popup_window` — всплывающие окна
* `onmouseover` — JS-события
* `right_clic` — отключена ли правая кнопка
* `login_form` — наличие формы входа
* `external_favicon` — внешний favicon
* `submit_email` — отправка данных на email
* `empty_title` — пустой title

Часто фишинг использует манипуляции интерфейсом.

---

## Ссылочная структура

* `nb_hyperlinks` — количество ссылок
* `ratio_intHyperlinks` — доля внутренних ссылок
* `ratio_extHyperlinks` — доля внешних
* `ratio_nullHyperlinks` — пустые ссылки
* `ratio_intRedirection`
* `ratio_extRedirection`

Фишинговые сайты часто ведут на внешние ресурсы.

---

## Бренд-маскировка

* `domain_in_brand`
* `brand_in_subdomain`
* `brand_in_path`
* `suspecious_tld`

## 4. Этапы выполнения проекта

### 4.1. Предобработка данных

* Проверка пропусков
* Проверка дубликатов
* Анализ распределения классов
* Кодирование категориальных признаков (если есть)
* Масштабирование признаков (при необходимости)

---

### 4.2. Разведочный анализ данных (EDA)

* Анализ баланса классов
* Корреляционная матрица
* Визуализация распределения признаков
* Анализ важности признаков

---

### 4.3. Построение моделей

Реализовать и сравнить несколько моделей:

1. Logistic Regression
2. Decision Tree
3. Random Forest
4. Gradient Boosting (например XGBoost / LightGBM)

---

### 4.4. Оценка моделей

Использовать метрики:

* Accuracy
* Precision
* Recall
* F1-score
* ROC-AUC

Основной акцент — **Recall**, так как важно минимизировать пропуск фишинговых сайтов.

Также:
* Построить Confusion Matrix
* Построить ROC-кривую

---

### 4.5. Работа с дисбалансом (если есть)

При необходимости:

* class_weight
* SMOTE
* undersampling

---

### 4.6. Подбор гиперпараметров

Использовать:

* GridSearchCV
  или
* RandomizedSearchCV

---

### 4.7. Финальная модель

* Выбрать лучшую модель
* Обучить на всех обучающих данных
* Сохранить модель (pickle / joblib)
